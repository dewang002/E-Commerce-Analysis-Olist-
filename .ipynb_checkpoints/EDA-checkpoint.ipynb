How did total orders, revenue, and average order value change over time?

```python
# 1
totalorders = pd.read_sql_query('''
SELECT COUNT(order_id) FROM olist_order_items_dataset
''', conn)
```

```python
totalorders
```

```python
orderovertime = pd.read_sql_query('''
SELECT DATE(shipping_limit_date) AS DATE, COUNT(order_id) as orders FROM olist_order_items_dataset
GROUP BY DATE(shipping_limit_date)
''', conn);
df = orderovertime
df.head()
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

sns.lineplot(x=df['DATE'], y=df['orders'], data=df, ax=ax)

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))

plt.xticks(rotation= 45) 
plt.tight_layout()
plt.show()
```

```python
# 2
totalPrice = pd.read_sql_query('''
SELECT sum(price) FROM olist_order_items_dataset
''', conn)
```

```python
totalPrice
```

```python
priceovertime = pd.read_sql_query('''
SELECT DATE(shipping_limit_date) AS DATE, SUM(PRICE) as spent FROM olist_order_items_dataset
GROUP BY DATE(shipping_limit_date)
''',conn)
df = priceovertime
df.head()
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

sns.lineplot(x=df['DATE'], y=df['spent'], data=df, ax=ax)

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))

plt.xticks(rotation= 45) 
plt.tight_layout()
plt.show()
```

```python
# 3
avgvalueovertime = pd.read_sql_query('''
SELECT DATE(shipping_limit_date) as DATE, avg(PRICE) as avgspent FROM olist_order_items_dataset
group by DATE(shipping_limit_date)
''',conn)
df = avgvalueovertime
df.head()
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

sns.lineplot(x=df['DATE'], y=df['avgspent'], data=df, ax=ax)

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))

plt.xticks(rotation= 45) 
plt.tight_layout()
plt.show()
```

```python
avgitmemovertime = pd.read_sql_query('''
SELECT order_item_id as ID, avg(PRICE) as itemavg FROM olist_order_items_dataset
group by order_item_id
order by itemavg
''',conn)
df = avgitmemovertime
df.head()
```

```python
sns.barplot(data=df, x=df['ID'], y=df['itemavg'], order=df['ID'])
```
